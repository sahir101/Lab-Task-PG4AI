```python
class Student:
    def __init__(self, name, age, course):
        self.name = name
        self.age = age
        self.course = course

    def display(self):
        print("Name:", self.name)
        print("Age:", self.age)
        print("Course:", self.course)


s1 = Student("sahir", 20, "AI")
s2 = Student("Sara", 21, "Python")

s1.display()
s2.display()
```

```python
class Graduate(Student):
    def __init__(self, name, age, course, research_topic):
        super().__init__(name, age, course)
        self.research_topic = research_topic

    def display(self):
        super().display()
        print("Research Topic:", self.research_topic)


g1 = Graduate("sahir", 19 , "Programming For AI", "Artificial Intelligence")

g1.display()
```
